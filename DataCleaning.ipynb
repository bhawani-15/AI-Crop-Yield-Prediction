```python
import pandas as pd
```

```python
df1 = pd.read_csv('Raw_Data/DES-District-Data-For-2016-17-to-2018-19.csv')
df2 = pd.read_csv('Raw_Data/DES-District-Data-For-2019-20-to-2021-22.csv')
df3 = pd.read_csv('Raw_Data/DES-District-Data-For-2022-23-to-2024-25.csv')
```

```python
print(df1.columns.tolist())
print(df2.columns.tolist())
print(df3.columns.tolist())
```

```python
print(df1.shape)
print(df2.shape)
print(df3.shape)
```

```python
print(df1.head())
print(df2.head())
print(df3.head())
```

```python
id_cols = ["State", "District", "Crop", "Season"]

for i, df in enumerate([df1, df2, df3], start=1):
    print(f"File {i}")
    print(df[id_cols].isna().sum())
    print()
```

```python
for i, df in enumerate([df1, df2, df3], start=1):
    print(f"File {i}")
    print(df[id_cols].dtypes)
    print()
```

```python
for i, df in enumerate([df1, df2, df3], start=1):
    duplicates = df.duplicated(subset=id_cols).sum()
    print(f"File {i}: {duplicates} duplicate records")
```

```python
id_cols = ["State", "District", "Crop", "Season"]

df = df1.merge(df2, on=id_cols, how="outer")
df = df.merge(df3, on=id_cols, how="outer")
```

```python
print(df.shape)
```

```python
print(df.columns.tolist())
```

```python
df["State"] = df["State"].astype(str).str.strip()
```

```python
print("Number of states:", df["State"].nunique())
print(sorted(df["State"].unique()))
```

```python
print(sorted(df["Crop"].dropna().unique()))
```

```python
aggregate_crops = [
    "Cereals",
    "Nutri/Coarse Cereals",
    "Other Pulses",
    "Total Food Grains",
    "Total Pulses"
]

df = df[
    ~df["Crop"].isin(aggregate_crops)
].copy()
```

```python
print("Number of crops:", df["Crop"].nunique())
print(sorted(df["Crop"].unique()))
```

```python
df = df[
    df["Season"].astype(str).str.strip().str.lower() != "total"
].copy()
```

```python
print("Seasons:")
print(sorted(df["Season"].unique()))
```

```python
df.to_csv("agricultural_yield_india_cleaned_wide.csv", index=False)

print("CSV saved successfully!")
```

```python
print(df.shape)

print("\nMissing values in each year:")
print(df.filter(like="Yield-").isna().sum())
```

```python
print("\nRows with missing values:")
print(df.isna().sum())
```

```python
id_cols = ["State", "District", "Crop", "Season"]

print(
    "Duplicate keys:",
    df.duplicated(subset=id_cols).sum()
)
```

```python
yield_cols = [
    col for col in df.columns
    if col.startswith("Yield-")
]

df_long = df.melt(
    id_vars=["State", "District", "Crop", "Season"],
    value_vars=yield_cols,
    var_name="Year",
    value_name="Yield_Kg_Ha"
)
```

```python
df_long["Year"] = (
    df_long["Year"]
    .str.replace("Yield-", "", regex=False)
    .str.strip()
)

df_long["Yield_Kg_Ha"] = pd.to_numeric(
    df_long["Yield_Kg_Ha"],
    errors="coerce"
)
```

```python
print(df_long.shape)
print(df_long.columns.tolist())
print(df_long.head(10))
```

```python
print("Total rows:", len(df_long))
print("Missing yield:", df_long["Yield_Kg_Ha"].isna().sum())
print(
    "Missing percentage:",
    round(df_long["Yield_Kg_Ha"].isna().mean() * 100, 2),
    "%"
)
```

```python
print(
    df_long.groupby("Year")["Yield_Kg_Ha"]
    .apply(lambda x: x.isna().sum())
)
```

```python
print(
    df_long.groupby("Crop")["Yield_Kg_Ha"]
    .apply(lambda x: x.isna().sum())
    .sort_values(ascending=False)
)
```

```python
print(
    df_long.groupby("State")["Yield_Kg_Ha"]
    .apply(lambda x: x.isna().sum())
    .sort_values(ascending=False)
)
```

```python
missing_by_state_crop = (
    df_long
    .groupby(["State", "Crop"])["Yield_Kg_Ha"]
    .apply(lambda x: x.isna().sum())
    .sort_values(ascending=False)
)

print(missing_by_state_crop.head(30))
```

```python
missing_by_season = (
    df_long
    .groupby("Season")["Yield_Kg_Ha"]
    .apply(lambda x: x.isna().sum())
)

print(missing_by_season)
```

```python
missing_percentage = (
    df_long.groupby(["State", "Crop"])["Yield_Kg_Ha"]
    .apply(lambda x: x.isna().mean() * 100)
    .sort_values(ascending=False)
)

print(missing_percentage.head(30))
```

```python
coverage = (
    df_long.groupby(["State", "Crop"])["Yield_Kg_Ha"]
    .agg(
        total="size",
        available="count"
    )
)

coverage["missing"] = coverage["total"] - coverage["available"]

coverage["available_percentage"] = (
    coverage["available"] / coverage["total"] * 100
)

print(coverage.sort_values("available_percentage").head(30))
```

```python
df_model = df_long.dropna(subset=["Yield_Kg_Ha"]).copy()
```

```python
print("Rows before:", len(df_long))
print("Rows after:", len(df_model))
print("Missing yield:", df_model["Yield_Kg_Ha"].isna().sum())
```

```python
print(df_model["Yield_Kg_Ha"].describe())
```

```python
df_model.to_csv(
    "agricultural_yield_india_cleaned_long.csv",
    index=False
)

print("CSV saved successfully!")
```
